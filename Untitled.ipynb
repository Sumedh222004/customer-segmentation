```python
import pandas as pd
import sqlite3

df = pd.read_csv("Retail_Data_Transactions.csv")
print(f"Rows: {len(df)}")
print(f"Columns: {list(df.columns)}")
print(df.head())
```

```python
# Convert date and explore the data
df["trans_date"] = pd.to_datetime(df["trans_date"], format="%d-%b-%y")

print(f"Date range: {df['trans_date'].min()} to {df['trans_date'].max()}")
print(f"Unique customers: {df['customer_id'].nunique()}")
print(f"Total revenue: ₹{df['tran_amount'].sum():,.0f}")
print(f"Avg transaction value: ₹{df['tran_amount'].mean():.2f}")
print(f"\nTransaction amount stats:")
print(df["tran_amount"].describe())
```

```python
# RFM Analysis
# Recency = days since last purchase
# Frequency = number of transactions
# Monetary = total spend

# Use the day after the last transaction as reference date
reference_date = df["trans_date"].max() + pd.Timedelta(days=1)

rfm = df.groupby("customer_id").agg(
    last_purchase=("trans_date", "max"),
    frequency=("trans_date", "count"),
    monetary=("tran_amount", "sum")
).reset_index()

rfm["recency"] = (reference_date - rfm["last_purchase"]).dt.days

rfm = rfm[["customer_id", "recency", "frequency", "monetary"]]

print(f"RFM table built for {len(rfm)} customers\n")
print(rfm.describe().round(2))
print("\nSample:")
print(rfm.head(10).to_string(index=False))
```

```python
# Score each customer 1-4 on each RFM dimension using NTILE logic
# Recency: lower days = better (score 4), higher days = worse (score 1)
# Frequency & Monetary: higher = better (score 4)

rfm["r_score"] = pd.qcut(rfm["recency"], q=4, labels=[4, 3, 2, 1]).astype(int)
rfm["f_score"] = pd.qcut(rfm["frequency"].rank(method="first"), q=4, labels=[1, 2, 3, 4]).astype(int)
rfm["m_score"] = pd.qcut(rfm["monetary"], q=4, labels=[1, 2, 3, 4]).astype(int)

rfm["rfm_score"] = rfm["r_score"].astype(str) + rfm["f_score"].astype(str) + rfm["m_score"].astype(str)

# Segment customers based on RFM scores
def segment_customer(row):
    r, f, m = row["r_score"], row["f_score"], row["m_score"]
    if r >= 3 and f >= 3 and m >= 3:
        return "Champions"
    elif r >= 3 and f >= 2:
        return "Loyal"
    elif r >= 3 and f <= 2:
        return "Promising"
    elif r == 2 and f >= 2:
        return "At Risk"
    elif r <= 2 and f <= 2 and m >= 3:
        return "High Value Churned"
    else:
        return "Lost"

rfm["segment"] = rfm.apply(segment_customer, axis=1)

# Segment summary
summary = rfm.groupby("segment").agg(
    customers=("customer_id", "count"),
    avg_recency=("recency", "mean"),
    avg_frequency=("frequency", "mean"),
    avg_monetary=("monetary", "mean"),
    total_revenue=("monetary", "sum")
).round(2).sort_values("total_revenue", ascending=False)

summary["revenue_pct"] = (summary["total_revenue"] / summary["total_revenue"].sum() * 100).round(1)

print("--- CUSTOMER SEGMENTATION RESULTS ---\n")
print(summary.to_string())
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle("Customer Segmentation & Revenue Intelligence", 
             fontsize=16, fontweight="bold")

colors = {
    "Champions": "#2ecc71",
    "Loyal": "#3498db", 
    "Promising": "#9b59b6",
    "At Risk": "#e67e22",
    "High Value Churned": "#e74c3c",
    "Lost": "#95a5a6"
}
segment_colors = [colors[s] for s in summary.index]

# Chart 1 — Revenue by Segment
ax1 = axes[0, 0]
bars = ax1.bar(summary.index, summary["total_revenue"], color=segment_colors)
ax1.set_title("Total Revenue by Segment")
ax1.set_ylabel("Revenue (₹)")
ax1.tick_params(axis="x", rotation=30)
for bar, pct in zip(bars, summary["revenue_pct"]):
    ax1.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 20000,
             f"{pct}%", ha="center", fontsize=9, fontweight="bold")

# Chart 2 — Customer Count by Segment
ax2 = axes[0, 1]
ax2.pie(summary["customers"], labels=summary.index, 
        colors=segment_colors, autopct="%1.1f%%", startangle=90)
ax2.set_title("Customer Distribution by Segment")

# Chart 3 — Avg Recency vs Avg Monetary (bubble = frequency)
ax3 = axes[1, 0]
for seg, row in summary.iterrows():
    ax3.scatter(row["avg_recency"], row["avg_monetary"],
                s=row["avg_frequency"] * 30,
                color=colors[seg], label=seg, alpha=0.8)
    ax3.annotate(seg, (row["avg_recency"], row["avg_monetary"]),
                 textcoords="offset points", xytext=(8, 4), fontsize=8)
ax3.set_xlabel("Avg Recency (days since last purchase)")
ax3.set_ylabel("Avg Monetary Value (₹)")
ax3.set_title("Recency vs Monetary\n(bubble size = avg frequency)")

# Chart 4 — Pareto: Cumulative Revenue
ax4 = axes[1, 1]
sorted_summary = summary.sort_values("total_revenue", ascending=False)
cumulative_pct = sorted_summary["revenue_pct"].cumsum()
ax4.bar(sorted_summary.index, sorted_summary["revenue_pct"], 
        color=[colors[s] for s in sorted_summary.index])
ax4_twin = ax4.twinx()
ax4_twin.plot(sorted_summary.index, cumulative_pct, 
              color="black", marker="o", linewidth=2)
ax4_twin.axhline(y=80, color="red", linestyle="--", alpha=0.7)
ax4_twin.set_ylabel("Cumulative Revenue %")
ax4.set_ylabel("Revenue %")
ax4.set_title("Pareto Analysis — Revenue Concentration")
ax4.tick_params(axis="x", rotation=30)

plt.tight_layout()
plt.savefig("segmentation_dashboard.png", dpi=150, bbox_inches="tight")
plt.show()
print("Dashboard saved.")
```

```python
# Save to SQLite and run advanced SQL queries
conn = sqlite3.connect("customers.db")
df.to_sql("transactions", conn, if_exists="replace", index=False)
rfm.to_sql("rfm_segments", conn, if_exists="replace", index=False)

print("Database created. Running SQL queries...\n")

# Query 1: Revenue by segment using SQL
q1 = pd.read_sql("""
    SELECT 
        segment,
        COUNT(DISTINCT r.customer_id) as customers,
        ROUND(SUM(r.monetary), 0) as total_revenue,
        ROUND(AVG(r.monetary), 0) as avg_revenue_per_customer,
        ROUND(SUM(r.monetary) * 100.0 / 
              (SELECT SUM(monetary) FROM rfm_segments), 1) as revenue_pct
    FROM rfm_segments r
    GROUP BY segment
    ORDER BY total_revenue DESC
""", conn)
print("--- SQL Query 1: Revenue by Segment ---")
print(q1.to_string(index=False))

# Query 2: Month over month revenue trend using SQL
q2 = pd.read_sql("""
    SELECT 
        strftime('%Y-%m', trans_date) as month,
        COUNT(DISTINCT customer_id) as unique_customers,
        COUNT(*) as transactions,
        ROUND(SUM(tran_amount), 0) as monthly_revenue,
        ROUND(AVG(tran_amount), 2) as avg_transaction
    FROM transactions
    GROUP BY month
    ORDER BY month
    LIMIT 12
""", conn)
print("\n--- SQL Query 2: Monthly Revenue Trend (first 12 months) ---")
print(q2.to_string(index=False))

# Query 3: High value churned customers — actionable list
q3 = pd.read_sql("""
    SELECT 
        customer_id,
        recency as days_since_purchase,
        frequency as total_transactions,
        monetary as total_spend,
        CASE 
            WHEN recency BETWEEN 100 AND 200 THEN 'Winback Priority 1'
            WHEN recency BETWEEN 200 AND 400 THEN 'Winback Priority 2'
            ELSE 'Likely Gone'
        END as winback_priority
    FROM rfm_segments
    WHERE segment = 'High Value Churned'
    ORDER BY monetary DESC
    LIMIT 10
""", conn)
print("\n--- SQL Query 3: High Value Churned — Winback List ---")
print(q3.to_string(index=False))

conn.close()
```
